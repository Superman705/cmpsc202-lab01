```python
import numpy as np
import time
import matplotlib.pyplot as plt
def baseline_max_sublist(arr):
    
    biggest_sum = -100

    for i in range(len(arr)):
        current_sum = 0
        for j in range(i, len(arr)):
            current_sum += arr[j]
            if current_sum > biggest_sum :
                biggest_sum = current_sum

            

    return (biggest_sum)

def kadane_max_sublist(arr):
    
    max_so_far = -10
    current_max = 0
    for i in range(len(arr)):
        current_max = current_max + arr[i]
        if current_max > max_so_far:
            max_so_far = current_max

        if current_max < 0:
            current_max = 0
    return(max_so_far)



def generate_data(n):
    data = []
    for _ in range((n)):
        i = np.random.randint(-100, 100)
        data.append(i)
    return data



sizes = [100, 500, 1000, 2500, 5000, 10000]
times = []
kadane_time = []

for i in range(len(sizes)):
    for j in range(10):
        data = generate_data(sizes[i])
        start = time.perf_counter()
        baseline_max_sublist(data)
        end = time.perf_counter()
        elapsed = end - start
        L = (sizes[i], elapsed)
        times.append(L)


average_times = []

for i in range(len(sizes)):
    size_times = 0
    for m in times:
        if sizes[i] == m[0]:
            size_times += m[1]
    average_time = size_times / 10
    average_times.append(average_time)
print(average_times)

for x in range(len(sizes)):
    for y in range(10):
        data = generate_data(sizes[x])
        start = time.perf_counter()
        kadane_max_sublist(data)
        end = time.perf_counter()
        elapsed = end - start
        L = (sizes[x], elapsed)
        kadane_time.append(L)

average_times_kadane = []

for i in range(len(sizes)):
    size_times_kadane = 0
    for m in kadane_time:
        if sizes[i] == m[0]:
            size_times_kadane += m[1]
    average_time_kadane = size_times_kadane / 10
    average_times_kadane.append(average_time_kadane)
print(f"average Kadane: {average_times_kadane}")

plt.plot(sizes, average_times, label="Baseline")
plt.plot(sizes, average_times_kadane, label="Kadane")

plt.xlabel("Array Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Maximum Subarray Algorithm Performance")

plt.legend()
plt.yscale('log')
plt.show()
```
